## 1 · Black-Scholes Pricing — Interactive

Black-Scholes gives the **fair value** of a European option assuming constant volatility, no dividends, and no arbitrage. The price depends on five parameters: S (spot price), K (strike), T (time to maturity), r (risk-free rate), and σ (implied volatility).

The left panel shows prices across all stock prices; the right panel shows the current call and put values at the chosen S. Notice that even an out-of-the-money option has positive value — that is **time value**: the probability that S crosses K before expiry.

```python
def plot_bs_pricing(S=100, K=100, T=1.0, r=0.05, sigma=0.20):
    clear_output(wait=True)
    S_range = np.linspace(50, 200, 300)
    calls   = black_scholes_call(S_range, K, T, r, sigma)
    puts    = black_scholes_put(S_range,  K, T, r, sigma)

    fig = make_subplots(rows=1, cols=2,
                        subplot_titles=["Option Prices vs Stock Price",
                                        "Call & Put at current S"])

    fig.add_trace(go.Scatter(x=S_range, y=calls, name="Call",
                             line=dict(color=COLORS["yellow"], width=2)), row=1, col=1)
    fig.add_trace(go.Scatter(x=S_range, y=puts, name="Put",
                             line=dict(color=COLORS["blue"], width=2)), row=1, col=1)
    fig.add_vline(x=S, line_dash="dash", line_color=COLORS["white"],
                  annotation_text=f"S={S}", row=1, col=1)
    fig.add_vline(x=K, line_dash="dot", line_color=COLORS["purple"],
                  annotation_text=f"K={K}", row=1, col=1)

    c = black_scholes_call(S, K, T, r, sigma)
    p = black_scholes_put(S,  K, T, r, sigma)
    fig.add_trace(go.Bar(
        x=["Call", "Put"], y=[c, p],
        marker_color=[COLORS["yellow"], COLORS["blue"]],
        text=[f"{c:.4f}", f"{p:.4f}"], textposition="outside",
        showlegend=False
    ), row=1, col=2)

    fig.update_layout(**LAYOUT, title_text="Black-Scholes Pricing")
    fig.show()
    pcp = put_call_parity_check(S, K, T, r, sigma)
    print(f"Put-call parity error: {pcp:.2e}  {'OK' if pcp < 1e-8 else 'FAIL'}")

widgets.interact(
    plot_bs_pricing,
    S     = widgets.FloatSlider(value=100,  min=50,   max=200,  step=1,    description="S"),
    K     = widgets.FloatSlider(value=100,  min=50,   max=200,  step=1,    description="K"),
    T     = widgets.FloatSlider(value=1.0,  min=0.05, max=3.0,  step=0.05, description="T (yrs)"),
    r     = widgets.FloatSlider(value=0.05, min=0.0,  max=0.15, step=0.01, description="r"),
    sigma = widgets.FloatSlider(value=0.20, min=0.05, max=0.80, step=0.01, description="sigma"),
);
```

## 2 · Greeks — Delta, Gamma, Vega

Greeks measure the sensitivity of the option price to each parameter. For a European call under Black-Scholes:

| Greek | Formula | Interpretation |
|-------|---------|----------------|
| **Delta** | $\Delta = N(d_1)$ | €-gain per €1 move in S — call ∈ [0,1], put ∈ [−1,0] |
| **Gamma** | $\Gamma = \dfrac{N'(d_1)}{S\,\sigma\sqrt{T}}$ | Rate of change of delta — identical for call and put |
| **Vega** | $\mathcal{V} = S\,N'(d_1)\sqrt{T}$ | €-gain per +1 pp in σ — identical for call and put |

**Key relationships:**

- Gamma and vega are proportional: $\mathcal{V} = S^2 \sigma T \cdot \Gamma$ — a high-gamma position is also high-vega
- Gamma and theta are linked by the BS PDE: $\Theta = -\frac{1}{2}\,\sigma^2 S^2\,\Gamma - r K e^{-rT} N(d_2)$ — the time decay you *pay* is the price of the convexity you *hold* (section 8)

**Reading the chart:** gamma spikes sharply ATM as T → 0. In practice this creates **pin risk**: near expiry, a small move around the strike causes delta to swing violently between 0 and 1, making the hedge nearly impossible to maintain.

```python
def plot_greeks(K=100, r=0.05, sigma=0.20):
    clear_output(wait=True)
    S_range    = np.linspace(50, 200, 300)
    maturities = [1.0, 0.5, 0.25, 0.1]
    palette    = [COLORS["blue"],   COLORS["orange"],
                  COLORS["yellow"], COLORS["pink"]]

    fig = make_subplots(rows=1, cols=3,
                        subplot_titles=["Delta", "Gamma", "Vega"])

    for T, col in zip(maturities, palette):
        deltas = black_scholes_delta(S_range, K, T, r, sigma)
        gammas = black_scholes_gamma(S_range, K, T, r, sigma)
        vegas  = black_scholes_vega( S_range, K, T, r, sigma)
        kw = dict(x=S_range, name=f"T={T}y", line=dict(color=col, width=2))
        fig.add_trace(go.Scatter(**kw, y=deltas), row=1, col=1)
        fig.add_trace(go.Scatter(**kw, y=gammas, showlegend=False), row=1, col=2)
        fig.add_trace(go.Scatter(**kw, y=vegas,  showlegend=False), row=1, col=3)

    for col in [1, 2, 3]:
        fig.add_vline(x=K, line_dash="dot", line_color=COLORS["white"], row=1, col=col)

    fig.update_layout(**LAYOUT, title_text="Black-Scholes Greeks")
    fig.show()

widgets.interact(
    plot_greeks,
    K     = widgets.FloatSlider(value=100,  min=50,   max=200,  step=5,    description="K"),
    r     = widgets.FloatSlider(value=0.05, min=0.0,  max=0.15, step=0.01, description="r"),
    sigma = widgets.FloatSlider(value=0.20, min=0.05, max=0.80, step=0.01, description="sigma"),
);
```

## 3 · Simulated GBM Price Paths

Under the **Geometric Brownian Motion** model, the stock follows:

$$dS = r \cdot S \, dt + \sigma \cdot S \, dW$$

where $dW$ is a Brownian increment. Paths are simulated under the **risk-neutral measure** (drift = r), which is the correct framework for option pricing and hedging. The dispersion of paths widens with σ and over time — this is the uncertainty the hedger must manage by continuously adjusting their delta position.

```python
def plot_paths(sigma=0.20, n_display=80):
    clear_output(wait=True)
    paths, tg = simulate_gbm_paths(
        S0=config.S0, r=config.r, sigma=sigma,
        T=config.T, n_steps=config.N_STEPS, n_paths=n_display, seed=0
    )
    fig = go.Figure()
    for i in range(n_display):
        fig.add_trace(go.Scatter(
            x=tg, y=paths[:, i], mode="lines",
            line=dict(width=0.7, color=COLORS["blue"]),
            opacity=0.3, showlegend=False
        ))
    fig.add_hline(y=config.K, line_dash="dash", line_color=COLORS["yellow"],
                  annotation_text=f"Strike K={config.K}",
                  annotation_font_color=COLORS["yellow"])
    fig.update_layout(
        **LAYOUT,
        title=f"GBM Paths (sigma={sigma:.0%}, {n_display} paths)",
        xaxis_title="Time (years)", yaxis_title="Stock Price",
    )
    fig.show()

widgets.interact(
    plot_paths,
    sigma     = widgets.FloatSlider(value=0.20, min=0.05, max=0.80, step=0.05, description="sigma"),
    n_display = widgets.IntSlider(  value=80,   min=10,   max=200,  step=10,   description="# paths"),
);
```

## 4 · P&L Distribution — Hedging Performance

The hedger sells the option, collects the premium, and rebalances the delta position at each step. In theory (continuous rebalancing, exact σ), the final P&L is **exactly zero** on every path. With discrete rebalancing, a residual error remains.

Two key properties to verify:
- **Mean P&L ≈ 0**: the hedge is unbiased — this is the martingale property of BS pricing
- **Hedging error** = std(P&L) / option price: shrinks as n_steps increases (see section 5)

Use the sliders to see the effect of transaction costs and option type (call vs put).

```python
def plot_pnl(n_steps=252, transaction_cost=0.0, option_type="call"):
    clear_output(wait=True)
    paths, tg = simulate_gbm_paths(
        S0=config.S0, r=config.r, sigma=config.SIGMA,
        T=config.T, n_steps=n_steps, n_paths=config.N_PATHS, seed=42
    )
    pnl = run_delta_hedge(paths, tg, config.K, config.r, config.SIGMA,
                          transaction_cost=transaction_cost,
                          option_type=option_type)

    price = (black_scholes_call if option_type == "call" else black_scholes_put)(
        config.S0, config.K, config.T, config.r, config.SIGMA
    )
    error = np.std(pnl) / price

    fig = go.Figure()
    fig.add_trace(go.Histogram(
        x=pnl, nbinsx=80,
        marker_color=COLORS["blue"], opacity=0.85,
        name="P&L"
    ))
    fig.add_vline(x=0, line_dash="dash", line_color=COLORS["red"],
                  annotation_text="Break-even",
                  annotation_font_color=COLORS["red"])
    fig.update_layout(
        **LAYOUT,
        title=f"{option_type.capitalize()} hedge | n_steps={n_steps} | "
              f"tx={transaction_cost:.3f} | Error: {error:.2%}",
        xaxis_title="Final P&L", showlegend=False,
    )
    fig.show()
    print(f"Mean: {pnl.mean():.4f}  |  Std: {pnl.std():.4f}  |  Error: {error:.2%}")

widgets.interact(
    plot_pnl,
    n_steps          = widgets.IntSlider(  value=252, min=4,   max=504,  step=4,    description="n_steps"),
    transaction_cost = widgets.FloatSlider(value=0.0, min=0.0, max=0.01, step=0.001,description="tx cost"),
    option_type      = widgets.Dropdown(   value="call", options=["call", "put"],    description="type"),
);
```

## 5 · Hedging Error vs Rebalancing Frequency

Expected relationship: error ∝ 1/√n_steps

**Intuition**: at each step, the unhedged gamma exposure over interval dt produces a P&L of order ½·Γ·(ΔS)². Since (ΔS)² ∝ dt under GBM, each step contributes a variance of order dt². Summing over n = T/dt steps, the total variance scales as n · dt² = T · dt, which goes to zero as dt → 0 (i.e., n → ∞). Taking the square root gives std(P&L) ∝ √dt = √(T/n) ∝ 1/√n.

```python
steps  = [4, 12, 52, 126, 252, 504]
labels = ["Quarterly", "Monthly", "Weekly", "Bi-weekly", "Daily", "Twice daily"]

df_freq = frequency_analysis(
    simulate_fn=simulate_gbm_paths,
    hedge_fn=run_delta_hedge,
    S0=config.S0, K=config.K, r=config.r, sigma=config.SIGMA, T=config.T,
    n_steps_range=steps, n_paths=500
)

# Theoretical 1/sqrt(n) fit
c_fit = df_freq["hedging_error"].iloc[0] * np.sqrt(df_freq["n_steps"].iloc[0])
n_fit = np.linspace(4, 504, 200)

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df_freq["n_steps"], y=df_freq["hedging_error"],
    mode="lines+markers", name="Simulated",
    line=dict(color=COLORS["blue"], width=2),
    marker=dict(size=9, color=COLORS["blue"]),
    text=labels, hovertemplate="%{text}<br>Error: %{y:.2%}"
))
fig.add_trace(go.Scatter(
    x=n_fit, y=c_fit / np.sqrt(n_fit),
    mode="lines", name="Theoretical",
    line=dict(color=COLORS["yellow"], dash="dash", width=1.5)
))
fig.update_layout(
    **LAYOUT,
    title="Hedging Error vs Rebalancing Frequency",
    xaxis_title="Number of rebalancing steps",
    yaxis_title="Hedging error", yaxis_tickformat=".1%",
)
fig.show()
```

## 6 · Volatility Mismatch Analysis

What happens if we hedge with the wrong σ? (Vega risk)

The P&L of a delta hedge depends on the gap between **realized volatility** (σ_real, used to simulate paths) and **implied volatility** (σ_hedge, used to compute deltas and option price):

- σ_hedge > σ_real → over-hedging → **positive mean P&L** (the premium charged was too high)
- σ_hedge < σ_real → under-hedging → **negative mean P&L** (the premium charged was too low)
- σ_hedge = σ_real → zero mean P&L (perfect calibration)

This is exactly the **vega risk** that market makers face when their vol model is miscalibrated.

```python
from analysis.metrics import vol_mismatch_analysis

paths_vm, tg_vm = simulate_gbm_paths(
    S0=config.S0, r=config.r, sigma=config.SIGMA,
    T=config.T, n_steps=config.N_STEPS, n_paths=config.N_PATHS, seed=42
)

sigma_range = np.round(np.arange(0.10, 0.40, 0.02), 3)
df_vm = vol_mismatch_analysis(
    run_hedge_fn=run_delta_hedge,
    paths=paths_vm, time_grid=tg_vm,
    K=config.K, r=config.r,
    sigma_real=config.SIGMA,
    sigma_hedge_range=sigma_range
)

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df_vm["sigma_hedge"], y=df_vm["mean_pnl"],
    mode="lines+markers", name="Mean P&L",
    line=dict(color=COLORS["blue"], width=2),
    marker=dict(size=8, color=COLORS["blue"]),
    fill="tozeroy", fillcolor="rgba(0,176,255,0.08)"
))
fig.add_vline(x=config.SIGMA, line_dash="dash", line_color=COLORS["yellow"],
              annotation_text=f"σ_real={config.SIGMA}",
              annotation_font_color=COLORS["yellow"])
fig.add_hline(y=0, line_dash="dot", line_color=COLORS["white"])
fig.update_layout(
    **LAYOUT,
    title="Mean P&L vs Hedge Volatility (Volatility Mismatch)",
    xaxis_title="σ used for hedging", xaxis_tickformat=".0%",
    yaxis_title="Mean P&L",
)
fig.show()
```

## 7 · Delta vs Gamma Hedging Comparison

Delta hedging leaves a residual **gamma exposure**: between two rebalancings, delta drifts as S moves, creating a tracking error proportional to (ΔS)².

**Gamma hedging** adds a second option (strike K_hedge ≠ K) to neutralize Γ simultaneously with Δ. The portfolio then remains closer to zero between rebalancings, achieving a lower hedging error at the same frequency.

Trade-off: it requires a second liquid instrument, more complex position sizing, and two bid-ask spreads to cross — in practice, gamma hedging is most useful when rebalancing is costly or infrequent.

```python
from hedging.gamma_hedge import run_gamma_hedge

paths_cmp, tg_cmp = simulate_gbm_paths(
    S0=config.S0, r=config.r, sigma=config.SIGMA,
    T=config.T, n_steps=52, n_paths=config.N_PATHS, seed=42
)

pnl_delta = run_delta_hedge(paths_cmp, tg_cmp, config.K, config.r, config.SIGMA)
pnl_gamma = run_gamma_hedge(paths_cmp, tg_cmp, config.K, config.r, config.SIGMA, K_hedge=110)

price = black_scholes_call(config.S0, config.K, config.T, config.r, config.SIGMA)

fig = go.Figure()
for pnl, label, col in [
    (pnl_delta, "Delta hedge",       COLORS["blue"]),
    (pnl_gamma, "D-Gamma hedge", COLORS["orange"]),
]:
    error = np.std(pnl) / price
    fig.add_trace(go.Histogram(
        x=pnl, name=f"{label} (error={error:.2%})",
        opacity=0.70, nbinsx=60, marker_color=col
    ))

fig.update_layout(
    **LAYOUT,
    barmode="overlay",
    title="P&L Distribution: Delta vs Delta-Gamma Hedging (weekly, n=52)",
    xaxis_title="Final P&L",
)
fig.show()
```

## 8 · P&L Attribution — Gamma P&L vs Theta P&L

For a delta-hedged portfolio, at each time step:

- **Gamma P&L** = ½·Γ·(ΔS)² — profit from realized price moves (convexity)
- **Theta P&L** = −½·Γ·σ²·S²·dt — cost of time decay (holding the option)

In expectation under Black-Scholes: Gamma P&L + Theta P&L ≈ 0.

> Key insight: when σ_real > σ_hedge → Gamma P&L > |Theta P&L| → net gain.

```python
def plot_pnl_attribution(sigma_real=0.20, sigma_hedge=0.20, n_steps=252):
    clear_output(wait=True)
    paths, tg = simulate_gbm_paths(
        S0=config.S0, r=config.r, sigma=sigma_real,
        T=config.T, n_steps=n_steps, n_paths=config.N_PATHS, seed=42
    )
    pnl, gamma_pnl, theta_pnl = run_delta_hedge(
        paths, tg, config.K, config.r, sigma_real,
        sigma_hedge=sigma_hedge, return_attribution=True
    )

    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=["P&L Attribution (mean over paths)",
                        "Total P&L distribution"]
    )
    fig.add_trace(go.Bar(
        x=["Gamma P&L", "Theta P&L", "Total P&L"],
        y=[gamma_pnl.mean(), theta_pnl.mean(), pnl.mean()],
        marker_color=[COLORS["yellow"], COLORS["red"], COLORS["blue"]],
        text=[f"{gamma_pnl.mean():.4f}", f"{theta_pnl.mean():.4f}", f"{pnl.mean():.4f}"],
        textposition="outside", textfont=dict(color=COLORS["text"])
    ), row=1, col=1)
    fig.add_trace(go.Histogram(
        x=pnl, nbinsx=60, marker_color=COLORS["blue"],
        opacity=0.80, name="Total P&L"
    ), row=1, col=2)
    fig.add_vline(x=0, line_dash="dash", line_color=COLORS["red"], row=1, col=2)

    mismatch = sigma_real - sigma_hedge
    fig.update_layout(
        **LAYOUT, showlegend=False,
        title_text=f"P&L Attribution | sigma_real={sigma_real:.0%}, sigma_hedge={sigma_hedge:.0%} "
                   f"| Mismatch={mismatch:+.0%}"
    )
    fig.show()
    print(f"Gamma P&L  : {gamma_pnl.mean():+.4f}  (std: {gamma_pnl.std():.4f})")
    print(f"Theta P&L  : {theta_pnl.mean():+.4f}  (std: {theta_pnl.std():.4f})")
    print(f"Total P&L  : {pnl.mean():+.4f}  (std: {pnl.std():.4f})")
    print(f"Gamma + Theta ~ Total ? {(gamma_pnl + theta_pnl).mean():+.4f}")

widgets.interact(
    plot_pnl_attribution,
    sigma_real  = widgets.FloatSlider(value=0.20, min=0.05, max=0.50, step=0.01, description="sigma_real"),
    sigma_hedge = widgets.FloatSlider(value=0.20, min=0.05, max=0.50, step=0.01, description="sigma_hedge"),
    n_steps     = widgets.IntSlider(  value=252,  min=12,   max=504,  step=12,   description="n_steps"),
);
```
